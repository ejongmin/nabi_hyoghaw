## 4. GPU 분류 실행 (cuDF)

핵심: 각 카테고리별 mega-regex → `cudf.Series.str.contains()` GPU 벡터 연산

```python
t_total = time.time()

# ── Step 1: 텍스트 컬럼 준비 (title 우선, 비어있으면 url fallback) ──
print("Preparing text column...")
titles = pdf["title"].fillna("").astype(str)
urls = pdf["url"].fillna("").astype(str)
empty_mask = titles.str.len() < 5
text_col = titles.copy()
text_col[empty_mask] = urls[empty_mask]
n_title = (~empty_mask).sum()
n_url = empty_mask.sum()
print(f"  From title: {n_title:,}  |  From url fallback: {n_url:,}")

# ── Step 2: cuDF Series로 변환 (GPU 메모리 전송) ──
print("Transferring to GPU...")
t0 = time.time()
gdf_text = cudf.Series(text_col.values)
print(f"  GPU transfer: {time.time()-t0:.1f}s  |  {len(gdf_text):,} rows on GPU")

# Case-insensitive: GPU에서 소문자 변환 (cuDF (?i) 미지원이므로)
print("  Converting to lowercase on GPU...")
t0 = time.time()
gdf_text = gdf_text.str.lower()
print(f"  Lowercase done in {time.time()-t0:.1f}s")

# ── Step 3: 카테고리별 mega-regex 매칭 ──
print("\n=== GPU regex matching (9 categories) ===")
S_MAX = 25.0
RTYPES = list(RISK_CONFIG.keys())
match_results = {}  # {rtype: cudf boolean Series}
avg_weights = {}

for rtype in RTYPES:
    terms = RISK_CONFIG[rtype]["terms"]
    # mega-regex: 모든 term 패턴을 OR로 합침
    mega = "|".join(f"(?:{pat.lower()})" for _, _, pat in terms)  # .lower() on patterns + text
    avg_w = sum(w for _, w, _ in terms) / len(terms)
    avg_weights[rtype] = avg_w

    t0 = time.time()
    try:
        hit = gdf_text.str.contains(mega, regex=True)
    except Exception as e:
        # cuDF regex가 일부 패턴 미지원 시 pandas fallback
        print(f"  {rtype:15s}: cuDF regex error, falling back to pandas...")
        hit_pd = text_col.str.contains(mega, case=False, na=False, regex=True)
        hit = cudf.Series(hit_pd.values)

    n_hit = int(hit.sum())
    elapsed = time.time() - t0
    match_results[rtype] = hit
    print(f"  {rtype:15s}: {n_hit:>10,} hits ({n_hit/len(gdf_text)*100:.2f}%) in {elapsed:.1f}s")

# ── Step 4: risk_types + severity 조합 (GPU → CPU) ──
print("\nBuilding risk_types and severity...")
t0 = time.time()

# GPU에서 boolean DataFrame 구성
match_pdf = pd.DataFrame({rtype: match_results[rtype].to_pandas() for rtype in RTYPES})

# severity: 매칭된 카테고리의 avg_weight 합산
severity_raw = pd.Series(0.0, index=range(len(pdf)))
for rtype in RTYPES:
    severity_raw += match_pdf[rtype].astype(float) * avg_weights[rtype]
severity = np.minimum(np.log(1 + severity_raw) / np.log(1 + S_MAX), 1.0).round(4)

# risk_types: 매칭된 카테고리 쉼표 연결
any_match = match_pdf.any(axis=1)
risk_types = pd.Series("other", index=range(len(pdf)))

# 매칭된 행만 문자열 조합 (소수이므로 빠름)
matched_idx = match_pdf.index[any_match]
print(f"  Matched rows: {len(matched_idx):,} / {len(pdf):,} ({len(matched_idx)/len(pdf)*100:.1f}%)")

def row_to_types(row):
    return ",".join(rt for rt in RTYPES if row[rt])

risk_types.loc[matched_idx] = match_pdf.loc[matched_idx].apply(row_to_types, axis=1)

print(f"  Combination done in {time.time()-t0:.1f}s")

# ── Step 5: 원본 DataFrame에 적용 ──
pdf["risk_types"] = risk_types.values
pdf["severity"] = severity.values

elapsed_total = time.time() - t_total
print(f"\n{'='*50}")
print(f"TOTAL elapsed: {elapsed_total:.1f}s ({elapsed_total/60:.1f}min)")
print(f"{'='*50}")
```

## 6. 저장 & 다운로드

```python
# ── entity_scores 타입 정규화 (ArrowInvalid 방지) ──
# 10M+ 행에 .apply()는 OOM 유발 → 청크 처리
import gc

print("Normalizing entity_scores (chunked)...")
def _force_float_list(x):
    if isinstance(x, (list, tuple, np.ndarray)):
        return [float(v) if not isinstance(v, dict) else float(v.get("score", 0.0)) for v in x]
    if x is None or (isinstance(x, float) and np.isnan(x)):
        return []
    return [float(x)] if isinstance(x, (int, float)) else []

CHUNK = 500_000
n = len(pdf)
for start in range(0, n, CHUNK):
    end = min(start + CHUNK, n)
    pdf.iloc[start:end, pdf.columns.get_loc("entity_scores")] = (
        pdf.iloc[start:end]["entity_scores"].apply(_force_float_list)
    )
    if (start // CHUNK) % 5 == 0:
        print(f"  {end:,} / {n:,} ({end/n*100:.0f}%)")
    gc.collect()

print(f"  Done: {n:,} rows normalized")

# ── event_time tz-naive 변환 ──
if "event_time" in pdf.columns:
    pdf["event_time"] = pd.to_datetime(pdf["event_time"], errors="coerce", utc=True)
    pdf["event_time"] = pdf["event_time"].dt.tz_localize(None)
    print("  event_time → tz-naive")

# ── 저장 ──
gc.collect()
print(f"Saving to {OUTPUT_PATH}...")
t0 = time.time()
pdf.to_parquet(OUTPUT_PATH, index=False)
print(f"Saved: {len(pdf):,} rows in {time.time()-t0:.1f}s")

# ── Drive에 저장했으면 로컬로도 다운로드 가능 ──
print("
다운로드 (로컬 PC로):")
try:
    files.download(OUTPUT_PATH)
except:
    print(f"자동 다운로드 실패 → Drive에서 직접 다운로드: {OUTPUT_PATH}")

print("
완료! 로컬에서 data/processed/risk_events.parquet로 교체하세요.")
```

## (참고) cuDF 미지원 시 CPU fallback

cuDF 설치 실패 또는 GPU 없는 환경에서는 아래 셀로 CPU pandas 실행 (느리지만 동작)

```python
# ── CPU Fallback (cuDF 없을 때만 실행) ──
# 이 셀은 위 Step 4가 실패했을 때만 실행하세요.

USE_CPU_FALLBACK = False  # ← True로 변경 후 실행

if USE_CPU_FALLBACK:
    print("Running CPU fallback (pandas str.contains)...")
    print("예상 소요: ~2-3시간")

    t_total = time.time()
    text_series = text_col  # Step 1에서 만든 것 재활용

    match_results_cpu = {}
    for rtype in RTYPES:
        terms = RISK_CONFIG[rtype]["terms"]
        mega = "|".join(f"(?:{pat})" for _, _, pat in terms)
        t0 = time.time()
        hit = text_series.str.contains(mega, case=False, na=False, regex=True)
        n_hit = hit.sum()
        elapsed = time.time() - t0
        match_results_cpu[rtype] = hit
        print(f"  {rtype:15s}: {n_hit:>10,} hits in {elapsed:.1f}s")

    # risk_types + severity 조합 (Step 4와 동일 로직)
    match_pdf = pd.DataFrame(match_results_cpu)
    severity_raw = pd.Series(0.0, index=range(len(pdf)))
    for rtype in RTYPES:
        avg_w = sum(w for _, w, _ in RISK_CONFIG[rtype]["terms"]) / len(RISK_CONFIG[rtype]["terms"])
        severity_raw += match_pdf[rtype].astype(float) * avg_w
    severity = np.minimum(np.log(1 + severity_raw) / np.log(1 + S_MAX), 1.0).round(4)

    any_match = match_pdf.any(axis=1)
    risk_types = pd.Series("other", index=range(len(pdf)))
    matched_idx = match_pdf.index[any_match]
    risk_types.loc[matched_idx] = match_pdf.loc[matched_idx].apply(row_to_types, axis=1)

    pdf["risk_types"] = risk_types.values
    pdf["severity"] = severity.values
    print(f"CPU fallback done in {time.time()-t_total:.0f}s")
else:
    print("CPU fallback disabled. Set USE_CPU_FALLBACK = True to enable.")
```
